Графики нужны инженеру ИИ не для презентаций, а чтобы контролировать процесс обучения моделей и исследовать распределение данных (датасетов) перед запуском обучения. Мы разберем два базовых графика, с которыми вы будете сталкиваться каждый день.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

### 1. Кривая обучения (Line Plot в Matplotlib)
Линейный график — это главный инструмент для отслеживания обучения нейросети. По оси X идут эпохи (время), по оси Y — ошибка (Loss). Мы смотрим, падает ли график: если падает, значит модель успешно учится.

```python
# Включаем стандартную тему оформления Seaborn (задает красивую сетку и фон)
sns.set_theme()

# Данные: 5 эпох обучения и падение ошибки (Loss)
epochs = [1, 2, 3, 4, 5]
loss_values = [0.95, 0.62, 0.38, 0.19, 0.08]

# 1. Создаем холст определенного размера (ширина 8, высота 4)
plt.figure(figsize=(8, 4))

# 2. Строим линию с точками (marker='o') красного цвета
plt.plot(epochs, loss_values, marker='o', color='red', label='Ошибка (Loss)')

# 3. Добавляем метаданные: заголовок и подписи осей
plt.title('Мониторинг обучения нейросети (Кривая потерь)', fontsize=14)
plt.xlabel('Эпоха обучения')
plt.ylabel('Величина ошибки')

# 4. Отображаем легенду и выводим график на экран
plt.legend()
plt.show()
```

### 2. Распределение данных (Histogram & KDE в Seaborn)
Перед подачей данных в ИИ важно понять, как распределены ваши признаки (например, возраст пользователей). Нет ли там аномальных перекосов? Для этого одной строчкой кода в Seaborn строится гистограмма плотности распределения.

```python
# Генерируем с помощью NumPy случайные данные, похожие на возраст 1000 клиентов (среднее 35 лет)
np.random.seed(42)
ages = np.random.normal(loc=35, scale=10, size=1000)

# Создаем холст
plt.figure(figsize=(8, 4))

# Строим гистограмму распределения сглаженной линией плотности (kde=True)
sns.histplot(ages, bins=30, kde=True, color='blue')

# Добавляем оформление
plt.title('Распределение возраста клиентов в датасете', fontsize=14)
plt.xlabel('Возраст (лет)')
plt.ylabel('Количество людей / Плотность')

plt.show()
```
