### Задача 1: Симуляция черно-белого изображения (Предобработка)
В компьютерном зрении картинка — это матрица, где каждое число означает яркость пикселя от 0 до 255.
**Задание:** Создать матрицу размера 4х4 со случайными целыми числами от 0 до 255 и нормализовать её (разделить на 255.0), чтобы значения стали в диапазоне от 0.0 до 1.0.

```python
# 1. Создаем матрицу 4х4 с числами от 0 до 255
image = np.random.randint(0, 256, size=(4, 4))
print("Исходное изображение:\n", image)

# 2. Нормализуем её (делим на 255.0)
normalized_image = image / 255.0
print("\nНормализованное изображение (от 0.0 до 1.0):\n", normalized_image)
```

### Задача 2: Подготовка данных для полносвязной нейросети (Reshaping)
На вход классической нейросети нужно подавать плоский вектор, а не квадратную картинку.
**Задание:** Возьми матрицу `image` (4х4) из первой задачи и "разверни" её в одномерный вектор размера 16.

```python
# Разворачиваем матрицу в плоский вектор размера 16 с помощью метода .reshape(-1)
flat_vector = image.reshape(-1)
print("Плоский вектор признаков (размерность", flat_vector.shape, "):\n", flat_vector)
```

### Задача 3: Фильтрация аномалий в данных (Маскирование)
Представь, что в векторе предсказаний нейросети появились ошибочные "выбросы" (числа меньше 0 или больше 1).
**Задание:** С помощью булевой маски выбери из вектора `bad_predictions` только те значения, которые строго находятся в правильном диапазоне: от 0 до 1 включительно.

```python
bad_predictions = np.array([-0.5, 0.1, 0.85, 1.2, 0.0, 2.5, 0.4])

# Отфильтровываем вектор с помощью булевой маски и оператора & (И)
good_predictions = bad_predictions[(bad_predictions >= 0) & (bad_predictions <= 1)]
print("Очищенные предсказания:", good_predictions)
```

### Задача 4: Расчет ошибки модели (Агрегация)
Ошибка MAE (Mean Absolute Error) считается как среднее абсолютное отклонение предсказаний от правильных ответов.
**Задание:** Посчитай среднюю абсолютную ошибку между `y_pred` (предсказания) и `y_true` (правильные ответы).

```python
y_true = np.array([10.0, 20.0, 30.0, 40.0])
y_pred = np.array([11.5, 19.0, 32.2, 38.5])

# Находим модуль разности через np.abs() и берем среднее через .mean()
mae = np.abs(y_pred - y_true).mean()
print("Средняя ошибка модели (MAE):", mae)
```

### Задача 5: Работа слоя нейросети (Матричное умножение)
Каждый слой нейросети делает одну базовую операцию: умножает матрицу входных данных (X) на матрицу весов (W).
**Задание:** Сделай строгое матричное умножение входного вектора `X` (размер 1х3) на матрицу весов `W` (размер 3х2).

```python
X = np.array([[0.5, 1.5, -0.2]])
W = np.array([
    [0.2, 0.8],
    [0.5, -0.1],
    [1.0, 0.3]
])

# Выполняем матричное умножение с помощью оператора @
output = X @ W
print("Выходной сигнал слоя нейросети (размерность", output.shape, "):\n", output)
```
